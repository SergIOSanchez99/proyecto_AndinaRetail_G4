<h1>Preprocesamiento</h1>

```python
def resumen_faltantes(df, nombre):
    faltantes = df.isnull().sum()
    porcentaje = (faltantes / len(df) * 100).round(2)
    resumen = pd.DataFrame({'faltantes': faltantes, '% faltante': porcentaje})
    resumen = resumen[resumen['faltantes'] > 0].sort_values('% faltante', ascending=False)
    print(f"\n--- {nombre} ---")
    print(resumen if not resumen.empty else "Sin valores faltantes")

resumen_faltantes(tiendas, 'tiendas')
resumen_faltantes(productos, 'productos')
resumen_faltantes(clientes, 'clientes')
resumen_faltantes(ventas, 'ventas')
resumen_faltantes(inventario, 'inventario')
```

<h1>Deteccion de Outliers</h1>

```python
def detectar_outliers_iqr(df, columna):
    Q1 = df[columna].quantile(0.25)
    Q3 = df[columna].quantile(0.75)
    IQR = Q3 - Q1
    limite_inferior = Q1 - 1.5 * IQR
    limite_superior = Q3 + 1.5 * IQR
    outliers = df[(df[columna] < limite_inferior) | (df[columna] > limite_superior)]
    print(f"{columna}: {len(outliers)} outliers ({len(outliers)/len(df)*100:.2f}%) "
          f"| límites: [{limite_inferior:.2f}, {limite_superior:.2f}]")
    return outliers

# Variables clave a revisar
print("--- VENTAS ---")
out_monto = detectar_outliers_iqr(ventas, 'monto_total')
out_cantidad = detectar_outliers_iqr(ventas, 'cantidad')
out_precio = detectar_outliers_iqr(ventas, 'precio_unitario')

print("\n--- CLIENTES ---")
out_edad = detectar_outliers_iqr(clientes, 'edad')
```

<h1>Medidas de Tendencia Central</h1>

```python
variables_clave = {
    'ventas': ['monto_total', 'cantidad', 'precio_unitario', 'descuento_pct'],
    'clientes': ['edad']
}

def estadistica_descriptiva(df, columna):
    return pd.Series({
        'media': df[columna].mean(),
        'mediana': df[columna].median(),
        'moda': df[columna].mode()[0],
        'desv_std': df[columna].std(),
        'varianza': df[columna].var(),
        'asimetria': df[columna].skew(),
        'curtosis': df[columna].kurt(),
        'min': df[columna].min(),
        'max': df[columna].max(),
    })

resumen_ventas = pd.DataFrame({col: estadistica_descriptiva(ventas, col) 
                                 for col in variables_clave['ventas']})
resumen_clientes = pd.DataFrame({col: estadistica_descriptiva(clientes, col) 
                                   for col in variables_clave['clientes']})

print("--- VENTAS ---")
display(resumen_ventas.round(2))
print("\n--- CLIENTES ---")
display(resumen_clientes.round(2))
```

<h1>HIPOTESIS</h1>

```python
from scipy import stats

# H0: no hay diferencia en monto_total entre Tienda y Web
# H1: sí hay diferencia
tienda = ventas[ventas['canal'] == 'Tienda']['monto_total']
web = ventas[ventas['canal'] == 'Web']['monto_total']

# Verificar supuesto de normalidad (con muestra, porque Shapiro no soporta N grande)
print("Shapiro Tienda (muestra 500):", stats.shapiro(tienda.sample(500, random_state=42)))
print("Shapiro Web (muestra 500):", stats.shapiro(web.sample(500, random_state=42)))

# Verificar supuesto de homogeneidad de varianzas (Levene)
levene_stat, levene_p = stats.levene(tienda, web)
print(f"\nLevene: stat={levene_stat:.3f}, p={levene_p:.4f}")

# Prueba t (Welch, no asume varianzas iguales - más robusta)
t_stat, p_value = stats.ttest_ind(tienda, web, equal_var=False)
print(f"\nT-test: t={t_stat:.3f}, p={p_value:.6f}")
print(f"Media Tienda: {tienda.mean():.2f} | Media Web: {web.mean():.2f}")
```

**Hipótesis 1: Ticket promedio entre canales (Tienda vs Web)**

Shapiro fuertemente rechazado (p ≈ 10⁻³⁸ en ambos grupos) → los datos NO son normales. Esto es esperado, ya que monto_total tiene skewness 8.89.
Levene rechazado (p < 0.0001) → varianzas distintas entre grupos, por eso usamos Welch (equal_var=False), correcto.
Resultado: t = -16.67, p < 0.000001 → se rechaza H0. Hay diferencia estadísticamente significativa: Web tiene ticket promedio mayor (S/571 vs S/464 en Tienda).

Sobre la violación de normalidad: no invalida la conclusión. Con N tan grande (147,463 en Tienda, 63,199 en Web), el Teorema del Límite Central garantiza que la distribución muestral de la media es aproximadamente normal, aunque los datos individuales no lo sean. 

```python
# Necesitamos unir ventas con clientes para tener la ciudad
ventas_cliente = ventas.merge(clientes[['id_cliente', 'ciudad']], on='id_cliente')

grupos_ciudad = [ventas_cliente[ventas_cliente['ciudad'] == c]['monto_total'] 
                  for c in ventas_cliente['ciudad'].unique()]

# Levene para homogeneidad de varianzas
levene_stat, levene_p = stats.levene(*grupos_ciudad)
print(f"Levene: stat={levene_stat:.3f}, p={levene_p:.4f}")

# ANOVA de una vía
f_stat, p_value = stats.f_oneway(*grupos_ciudad)
print(f"\nANOVA: F={f_stat:.3f}, p={p_value:.6f}")

print("\nMedia por ciudad:")
print(ventas_cliente.groupby('ciudad')['monto_total'].mean().round(2).sort_values(ascending=False))
```

**Hipótesis 2: ANOVA — monto por ciudad**

Levene no es rechazado (p=0.69) → varianzas homogéneas, supuesto cumplido.
F=0.571, p=0.683 → NO se rechaza H0. No hay diferencia significativa en monto_total entre ciudades (las medias van de 506 a 516, muy similares).

Esto sugiere que el monto de venta no depende de la ciudad, lo cual contrasta con lo que se espera encontrar en la Parte 2 sobre la caída de margen en Trujillo — ahí el problema no está en el monto de venta sino en el margen (descuento y costos), como lo indica el enunciado. 

```python
tabla_contingencia = pd.crosstab(ventas_prod['categoria'], ventas_prod['metodo_pago'])
print(tabla_contingencia)

chi2, p_value, dof, expected = stats.chi2_contingency(tabla_contingencia)
print(f"\nChi2={chi2:.3f}, p={p_value:.6f}, gl={dof}")

# Verificar supuesto: frecuencias esperadas >= 5
print("\n¿Todas las frecuencias esperadas >= 5?", (expected >= 5).all())
```

**Hipótesis 3: Chi-cuadrado — categoría vs método de pago**

Supuesto de frecuencias esperadas ≥5 cumplido.
Chi2=196.4, p<0.000001 → se rechaza H0. Hay asociación significativa entre categoría y método de pago (no son independientes).

Con gl=25 y un chi2 tan alto respecto a los grados de libertad, la asociación es estadísticamente muy fuerte

```python
# IC 95% para el ticket promedio general
media = ventas['monto_total'].mean()
sem = stats.sem(ventas['monto_total'])
ic = stats.t.interval(0.95, len(ventas)-1, loc=media, scale=sem)
print(f"IC 95% ticket promedio: [{ic[0]:.2f}, {ic[1]:.2f}]")

# IC 95% para tasa de descuento promedio
media_desc = ventas['descuento_pct'].mean()
sem_desc = stats.sem(ventas['descuento_pct'].dropna())
ic_desc = stats.t.interval(0.95, ventas['descuento_pct'].dropna().shape[0]-1, 
                            loc=media_desc, scale=sem_desc)
print(f"IC 95% descuento promedio: [{ic_desc[0]:.4f}, {ic_desc[1]:.4f}]")
```
